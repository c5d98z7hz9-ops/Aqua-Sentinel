# My Web App
This notebook is served as a web page by Voilà. Code cells are hidden; outputs and widgets are shown.

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display

slider = widgets.IntSlider(value=5, min=1, max=20, description='Waves:')
out = widgets.Output()

def draw(change=None):
    with out:
        out.clear_output(wait=True)
        x = np.linspace(0, 2*np.pi, 500)
        fig, ax = plt.subplots(figsize=(7,3))
        ax.plot(x, np.sin(slider.value * x))
        ax.set_title(f'sin({slider.value}x)')
        plt.show()

slider.observe(draw, names='value')
display(slider, out)
draw()
```
